```python
import pandas as pd
```

```python
records = [
    {"roll_number": "STU001", "name": "Aarav Sharma", "course": "Python Basics", "mark": 84},
    {"roll_number": "STU002", "name": "Diya Patel", "course": "Python Basics", "mark": 91},
    {"roll_number": "STU003", "name": "Kabir Singh", "course": "Data Analytics", "mark": 76},
]
```

```python
def add_record(records):
    
    print("\n--- Add student ---")
    
    roll_number = input("Roll number: ")
    name = input("Name: ")
    course = input("Course: ")
    mark = int(input("Mark: "))

    record = {
        "roll_number": roll_number,
        "name": name,
        "course": course,
        "mark": mark,
    }
    
    records.append(record)
    
    return "add_record executed"
```

```python
def view_records(records):
    print("===============================================")
    print("\n--- All students ---")
    df = pd.DataFrame(records)
    print(df)
    print("===============================================")
    return "view_records executed"
```

```python
def search_record(records):
    print("\n--- Search student ---")
    roll_number = input("Roll number: ")
    
    
    for record in records:
        if record["roll_number"] == roll_number:
            print("==============================")
            print(f"Name   : {record['name']}")
            print(f"Course : {record['course']}")
            print(f"Mark   : {record['mark']}")
            print("==============================")
            return "search_record executed"

    print("Student record not found.")
```

```python
def update_mark(records):
    print("\n--- Update mark ---")
    roll_number = input("Roll number: ")

    for record in records:
        if record["roll_number"] == roll_number:
            record["mark"] = int(input("Mark: "))
            return "Mark updated!"

    print("Student record not found.")
    return "update_mark executed"
```

```python
def show_menu():
    print("\nSTUDENT RECORD MANAGEMENT")
    print("1. Add student")
    print("2. View all students")
    print("3. Search by roll number")
    print("4. Update mark")
    print("5. Exit")
```

```python
def main():
        actions = {
        "1": add_record,
        "2": view_records,
        "3": search_record,
        "4": update_mark,
    }

    while True:
        show_menu()
        choice = input("Choose an option (1-5): ")

        if choice == "5":
            print("Goodbye.")
            break

        act = actions.get(choice)
        if act:
            print(act(records))
        else:
            print("Invalid menu option. Choose a number from 1 to 6")


if __name__ == "__main__":
    main()
```
